```python
import os, gzip, shutil, scanpy as sc
from pathlib import Path

SCRNA_DIR = "organized/scRNA"
OUTDIR = "processed/scRNA_h5ad"
Path(OUTDIR).mkdir(parents=True, exist_ok=True)

samples = sorted(set("_".join(f.split("_")[:4]) for f in os.listdir(SCRNA_DIR)))

for sample in samples:
    tmp = Path("tmp")/sample
    tmp.mkdir(parents=True, exist_ok=True)

    mapping = {
        f"{sample}_barcodes.tsv.gz":"barcodes.tsv.gz",
        f"{sample}_features.tsv.gz":"features.tsv.gz",
        f"{sample}_matrix.mtx.gz":"matrix.mtx.gz",
    }

    for src,dst in mapping.items():
        shutil.copy(Path(SCRNA_DIR)/src, tmp/dst)

    adata = sc.read_10x_mtx(tmp, var_names="gene_symbols")
    adata.write(Path(OUTDIR)/f"{sample}.h5ad")

print("Done")
```
