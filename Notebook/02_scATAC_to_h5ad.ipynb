```python
import os, gzip, pandas as pd, scipy.io, anndata as ad
from pathlib import Path

ATAC_DIR = "organized/scATAC"
OUTDIR = "processed/scATAC_h5ad"
Path(OUTDIR).mkdir(parents=True, exist_ok=True)

samples = sorted(set("_".join(f.split("_")[:4]) for f in os.listdir(ATAC_DIR)))

for sample in samples:
    X = scipy.io.mmread(f"{ATAC_DIR}/{sample}_matrix.mtx.gz").T.tocsr()
    obs = pd.read_csv(f"{ATAC_DIR}/{sample}_barcodes.tsv.gz", header=None, names=["barcode"])
    var = pd.read_csv(f"{ATAC_DIR}/{sample}_peaks.bed.gz", sep="\t", header=None,
                      names=["chr","start","end"])

    adata = ad.AnnData(X=X, obs=obs, var=var)
    adata.write(f"{OUTDIR}/{sample}.h5ad")

print("Done")
```
