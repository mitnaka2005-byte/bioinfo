```python
import gzip, pandas as pd, anndata as ad
from pathlib import Path
import numpy as np

INDIR = "organized/bulkRNA"
OUTDIR = "processed/bulkRNA_h5ad"
Path(OUTDIR).mkdir(parents=True, exist_ok=True)

for f in Path(INDIR).glob("*ReadsPerGene.txt.gz"):
    df = pd.read_csv(f, sep="\t", header=None)
    df = df.iloc[4:]  # remove STAR QC rows

    genes = df.iloc[:,0].astype(str)
    counts = pd.to_numeric(df.iloc[:,1], errors="coerce").fillna(0)

    adata = ad.AnnData(
        X=counts.to_numpy().reshape(1,-1),
        var=pd.DataFrame(index=genes),
        obs=pd.DataFrame(index=[f.stem])
    )

    adata.write(Path(OUTDIR)/(f.stem + ".h5ad"))

print("Done")
```
